### 1. Worked example

The dataset contains simulated quarterly Vietnamese macro series, 2005Q1–2024Q4: real GDP growth, CPI inflation, policy rate, credit growth (all % y/y), and the VN-Index quarterly return and level. In this session we model **inflation**.

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA
mac = pd.read_csv(DATA + "vn_macro_quarterly.csv")
mac.index = pd.PeriodIndex(mac["quarter"], freq="Q")
infl = mac["inflation"]
fig, ax = plt.subplots(1, 3, figsize=(14, 3))
infl.plot(ax=ax[0], title="Inflation (% y/y)"); plot_acf(infl, ax=ax[1], lags=16); plot_pacf(infl, ax=ax[2], lags=16, method="ywm")
plt.show()
```
