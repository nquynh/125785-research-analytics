### 1. Worked example

We model four variables jointly: GDP growth, inflation, the policy rate and credit growth. Question: **How do GDP and credit respond to a monetary-policy (policy-rate) shock?**

```python
from statsmodels.tsa.api import VAR
mac = pd.read_csv(DATA + "vn_macro_quarterly.csv")
mac.index = pd.PeriodIndex(mac["quarter"], freq="Q").to_timestamp()
y = mac[["gdp_growth", "inflation", "policy_rate", "credit_growth"]]
model = VAR(y)
print(model.select_order(6).summary())
```

```python
var = model.fit(2)               # lag length chosen in Exercise 9.1
print(var.summary())
```
