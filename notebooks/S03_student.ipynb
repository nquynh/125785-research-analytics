### 1. Worked example

We model firm profitability: roa = β0 + β1·leverage + β2·size + β3·sales_growth + β4·tangibility + β5·state_owned + industry and year dummies + u. In formulas, `C(industry)` creates dummy variables automatically.

```python
corp = pd.read_csv(DATA + "vn_corp_panel.csv").dropna(subset=["esg_score", "sales_growth"])
lo, hi = corp["roa"].quantile([0.01, 0.99]); corp["roa_w"] = corp["roa"].clip(lo, hi)
m1 = smf.ols("roa_w ~ leverage + size", data=corp).fit()
m2 = smf.ols("roa_w ~ leverage + size + sales_growth + tangibility + state_owned + C(industry) + C(year)", data=corp).fit()
print(m2.summary().tables[1])
```

```python
# A compact comparison table of the two models
from statsmodels.iolib.summary2 import summary_col
print(summary_col([m1, m2], stars=True, model_names=["(1) Basic", "(2) Full"],
                  regressor_order=["leverage", "size", "sales_growth", "tangibility", "state_owned"],
                  drop_omitted=True, info_dict={"N": lambda x: f"{int(x.nobs)}", "Adj. R2": lambda x: f"{x.rsquared_adj:.3f}"}))
```
