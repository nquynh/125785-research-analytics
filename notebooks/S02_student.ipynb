### 1. Worked example

**The dataset.** Monthly returns for 80 simulated HOSE stocks, Jan 2015–Dec 2024, with the risk-free rate `rf`, the market excess return `mkt_rf`, and size (`smb`) and value (`hml`) factors. The CAPM regression is: excess return_i = α + β · mkt_rf + u.

```python
stk = pd.read_csv(DATA + "vn_stock_returns_monthly.csv")
stk["ex_ret"] = stk["ret"] - stk["rf"]          # stock excess return
s01 = stk[stk["ticker"] == "S01"].copy()
capm = smf.ols("ex_ret ~ mkt_rf", data=s01).fit()
print(capm.summary())
```

```python
plt.scatter(s01["mkt_rf"], s01["ex_ret"], s=12)
x = np.linspace(s01["mkt_rf"].min(), s01["mkt_rf"].max(), 50)
plt.plot(x, capm.params["Intercept"] + capm.params["mkt_rf"] * x, color="red")
plt.xlabel("Market excess return"); plt.ylabel("S01 excess return"); plt.show()
```
