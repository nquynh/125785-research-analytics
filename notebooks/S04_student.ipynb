### 1. Worked example

Research question for this session (a green-finance example): **Is ESG disclosure associated with a lower cost of debt?** Model: cost_of_debt = β0 + β1·esg_score + β2·size + β3·leverage + β4·roa + year dummies + u.

```python
corp = pd.read_csv(DATA + "vn_corp_panel.csv").dropna(subset=["esg_score", "sales_growth"])
kd = smf.ols("cost_of_debt ~ esg_score + size + leverage + roa + C(year)", data=corp).fit()
print(kd.summary().tables[1])
```

```python
# Residual plot
plt.scatter(kd.fittedvalues, kd.resid, s=8); plt.axhline(0, color="red")
plt.xlabel("Fitted values"); plt.ylabel("Residuals"); plt.title("Residuals vs fitted"); plt.show()
```
