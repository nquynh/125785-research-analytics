### 1. Worked example

Research question: **What determines whether a Vietnamese listed firm pays a cash dividend?** Outcome `div_payer` (1/0). Explanatory variables: roa, leverage, size, state_owned.

```python
corp = pd.read_csv(DATA + "vn_corp_panel.csv").dropna(subset=["esg_score", "sales_growth"])
lo, hi = corp["roa"].quantile([0.01, 0.99]); corp["roa_w"] = corp["roa"].clip(lo, hi)
print("Share of dividend payers:", round(corp["div_payer"].mean(), 3))
f = "div_payer ~ roa_w + leverage + size + state_owned"
lpm = smf.ols(f, data=corp).fit(cov_type="cluster", cov_kwds={"groups": pd.factorize(corp["firm_id"])[0]})
logit = smf.logit(f, data=corp).fit(disp=0)
probit = smf.probit(f, data=corp).fit(disp=0)
print(logit.summary().tables[1])
```

```python
# Average marginal effects (AME): the change in probability for a one-unit change in x
print(logit.get_margeff(at="overall").summary())
```
