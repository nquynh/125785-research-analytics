### 1. Worked example

Question: **What is the effect of leverage on ROA?** Leverage is endogenous: unobserved firm quality raises ROA and lowers borrowing. We compare (1) pooled OLS, (2) firm fixed effects, and (3) 2SLS using `ind_leverage` (the average leverage of *other* firms in the same industry and year) as an instrument.

```python
from linearmodels.iv import IV2SLS
corp = pd.read_csv(DATA + "vn_corp_panel.csv").dropna(subset=["esg_score", "sales_growth"])
lo, hi = corp["roa"].quantile([0.01, 0.99]); corp["roa_w"] = corp["roa"].clip(lo, hi)
corp["const"] = 1
ols = IV2SLS(corp["roa_w"], corp[["const", "size", "sales_growth"]], None, None).fit(cov_type="clustered", clusters=corp["firm_id"])
ols_l = IV2SLS(corp["roa_w"], corp[["const", "leverage", "size", "sales_growth"]], None, None).fit(cov_type="clustered", clusters=corp["firm_id"])
iv = IV2SLS(corp["roa_w"], corp[["const", "size", "sales_growth"]], corp["leverage"], corp["ind_leverage"]).fit(cov_type="clustered", clusters=corp["firm_id"])
print(iv.summary.tables[1])
```
