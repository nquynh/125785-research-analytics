### 1. Worked example

Can our models forecast **inflation** better than a naive random walk? We estimate on 2005Q1–2018Q4 and evaluate one-quarter-ahead forecasts on 2019Q1–2024Q4 (24 quarters), re-estimating each quarter (expanding window).

```python
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.api import VAR
mac = pd.read_csv(DATA + "vn_macro_quarterly.csv")
mac.index = pd.PeriodIndex(mac["quarter"], freq="Q")
y = mac[["gdp_growth", "inflation", "policy_rate", "credit_growth"]]
test_idx = y.loc["2019Q1":].index
fc = {"naive": [], "AR(1)": [], "VAR(2)": []}
for t in test_idx:
    train = y.loc[:t - 1]
    fc["naive"].append(train["inflation"].iloc[-1])
    ar = ARIMA(train["inflation"].values, order=(1, 0, 0)).fit()
    fc["AR(1)"].append(ar.forecast(1)[0])
    v = VAR(train.values).fit(2)
    fc["VAR(2)"].append(v.forecast(train.values[-2:], 1)[0, 1])
fc = pd.DataFrame(fc, index=test_idx); fc["actual"] = y.loc[test_idx, "inflation"]
fc.head()
```
