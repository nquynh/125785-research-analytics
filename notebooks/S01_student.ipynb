### 1. Worked example

**The dataset.** `vn_corp_panel.csv` is a simulated panel of 150 non-financial firms listed in Vietnam, 2015–2024 (1,500 firm-years).
```python
corp = pd.read_csv(DATA + "vn_corp_panel.csv")
print(corp.shape)
corp.head()
```
